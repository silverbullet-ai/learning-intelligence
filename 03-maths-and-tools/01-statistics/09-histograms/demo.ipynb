# Histograms

This notebook demonstrates:

- Creating histograms
- Understanding bins
- Changing bin sizes
- Frequency distributions
- Histogram vs. bar chart
- Distribution visualization
- KDE intuition

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ages = [23, 24, 25, 30, 34, 36, 40, 50, 60, 75, 80]

print("Ages:", ages)
```

```python
plt.hist(
    ages,
    bins=8,
    edgecolor="black"
)

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Histogram of Ages")

plt.show()
```

```python
plt.hist(
    ages,
    bins=8,
    edgecolor="black"
)

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Histogram of Ages")

plt.show()
```

```python
frequency, bin_edges = np.histogram(ages, bins=8)

frequency_table = pd.DataFrame({
    "Bin Start": bin_edges[:-1],
    "Bin End": bin_edges[1:],
    "Frequency": frequency
})

frequency_table
```

```python
np.random.seed(42)

heights = np.random.normal(
    loc=170,
    scale=8,
    size=1000
)

plt.hist(
    heights,
    bins=20,
    edgecolor="black"
)

plt.xlabel("Height (cm)")
plt.ylabel("Frequency")
plt.title("Histogram of Heights")

plt.show()
```

```python
categories = ["Python", "Java", "C++", "JavaScript"]
students = [40, 25, 15, 20]

plt.bar(
    categories,
    students
)

plt.xlabel("Programming Language")
plt.ylabel("Number of Students")
plt.title("Bar Chart Example")

plt.show()
```

```python
from scipy.stats import gaussian_kde

density = gaussian_kde(heights)

x = np.linspace(
    heights.min(),
    heights.max(),
    300
)

plt.hist(
    heights,
    bins=20,
    density=True,
    alpha=0.5,
    edgecolor="black"
)

plt.plot(
    x,
    density(x)
)

plt.xlabel("Height (cm)")
plt.ylabel("Density")
plt.title("Histogram with KDE")

plt.show()
```

## Summary

A histogram:

- Groups numerical observations into bins.
- Shows the frequency distribution.
- Helps visualize the shape and spread of data.
- Is sensitive to the chosen bin size.
- Provides intuition for probability distributions.

KDE provides a smooth estimate of the underlying data distribution and is useful for understanding the connection between empirical data and probability density.
