# Percentiles and Quartiles

This notebook demonstrates:

- Percentage vs Percentile
- Percentile calculation
- Quartiles
- Q1, Q2, and Q3
- Median and Q2
- Interquartile Range (IQR)
- Box Plot

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = np.array([
    2, 2, 3, 4, 5, 5, 6,
    7, 8, 8, 8, 9, 9, 10
])

print("Dataset:")
print(data)

print("\nNumber of observations:", len(data))
```

```python
numbers = np.array([1, 2, 3, 4, 5, 6])

odd_count = np.sum(numbers % 2 != 0)
total_count = len(numbers)

percentage = (odd_count / total_count) * 100

print("Odd numbers:", odd_count)
print("Total numbers:", total_count)
print("Percentage of odd numbers:", percentage, "%")
```

```python
numbers = np.array([1, 2, 3, 4, 5, 6])

odd_count = np.sum(numbers % 2 != 0)
total_count = len(numbers)

percentage = (odd_count / total_count) * 100

print("Odd numbers:", odd_count)
print("Total numbers:", total_count)
print("Percentage of odd numbers:", percentage, "%")
```

```python
Q1 = np.percentile(data, 25)
Q2 = np.percentile(data, 50)
Q3 = np.percentile(data, 75)

quartiles = pd.DataFrame({
    "Quartile": ["Q1", "Q2", "Q3"],
    "Percentile": [25, 50, 75],
    "Value": [Q1, Q2, Q3]
})

quartiles
```

```python
median = np.median(data)

print("Median:", median)
print("Q2:", Q2)

print("Are Median and Q2 equal?", median == Q2)
```

```python
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
plt.boxplot(data)

plt.ylabel("Value")
plt.title("Box Plot Showing Quartiles")

plt.show()
```

```python
plt.hist(
    data,
    bins=8,
    edgecolor="black"
)

plt.axvline(
    Q1,
    linestyle="--",
    label="Q1 (25th Percentile)"
)

plt.axvline(
    Q2,
    linestyle="--",
    label="Q2 (Median)"
)

plt.axvline(
    Q3,
    linestyle="--",
    label="Q3 (75th Percentile)"
)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Percentiles and Quartiles")

plt.legend()
plt.show()
```

```python
series = pd.Series(data)

print("Q1:", series.quantile(0.25))
print("Q2:", series.quantile(0.50))
print("Q3:", series.quantile(0.75))
```

## Summary

- Q1 represents the 25th percentile.
- Q2 represents the 50th percentile and is the median.
- Q3 represents the 75th percentile.
- IQR measures the spread of the middle 50% of the data.

IQR = Q3 - Q1
