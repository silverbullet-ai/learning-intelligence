# Random Variables

This notebook demonstrates:

- Random Variables
- Discrete Random Variables
- Continuous Random Variables
- Coin Toss Simulation
- Dice Roll Simulation

```python
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
coin = random.choice(["Head", "Tail"])

X = 0 if coin == "Head" else 1

print("Outcome:", coin)
print("Random Variable X:", X)
```

```python
coin_results = [random.choice(["Head", "Tail"]) for _ in range(20)]

random_variable = [
    0 if toss == "Head" else 1
    for toss in coin_results
]

df = pd.DataFrame({
    "Outcome": coin_results,
    "X": random_variable
})

df
```

```python
coin_results = [random.choice(["Head", "Tail"]) for _ in range(20)]

random_variable = [
    0 if toss == "Head" else 1
    for toss in coin_results
]

df = pd.DataFrame({
    "Outcome": coin_results,
    "X": random_variable
})

df
```

```python
roll = random.randint(1, 6)

print("Dice Roll:", roll)
```

```python
dice = np.random.randint(1, 7, size=500)

plt.hist(
    dice,
    bins=np.arange(1, 8) - 0.5,
    edgecolor="black"
)

plt.title("Discrete Random Variable - Dice Rolls")
plt.xlabel("Dice Value")
plt.ylabel("Frequency")

plt.xticks(range(1, 7))

plt.show()
```

```python
heights = np.random.normal(
    loc=170,
    scale=8,
    size=1000
)

plt.hist(
    heights,
    bins=25,
    edgecolor="black"
)

plt.title("Continuous Random Variable - Heights")
plt.xlabel("Height (cm)")
plt.ylabel("Frequency")

plt.show()
```

## Summary

- Coin Toss → Discrete Random Variable
- Dice Roll → Discrete Random Variable
- Heights → Continuous Random Variable

Random Variables convert the outcomes of random experiments into numerical values that can be analyzed statistically.
