````python
import pandas as pd

data = {
    "hours_studied": [2, 3, 4, 5, 6],
    "exam_score": [50, 60, 70, 80, 90]
}

df = pd.DataFrame(data)

print(df.cov())
print(df.corr())
````

````python
spearman = df["hours_studied"].corr(
    df["exam_score"],
    method="spearman"
)

print(spearman)
````

````python
import matplotlib.pyplot as plt

plt.scatter(
    df["hours_studied"],
    df["exam_score"]
)

plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.title("Hours Studied vs Exam Score")

plt.show()
````

## Workflow

```txt
Two Variables  
      ↓  
Explore the Data  
      ↓  
Create Scatter Plot  
      ↓  
Calculate Covariance  
      ↓  
Calculate Correlation  
      ↓  
Interpret Direction and Strength  
      ↓  
Use Results for EDA / Feature Analysis  
```

## Key Takeaways

- Covariance measures how two variables change together.
- Positive covariance indicates movement in the same direction.
- Negative covariance indicates movement in opposite directions.
- Covariance has an unbounded range.
- Covariance depends on the scale and units of the variables.
- Correlation normalizes the relationship to a range of -1 to +1.
- Pearson correlation measures linear relationships.
- Spearman correlation measures monotonic relationships using ranks.
- Correlation is commonly used in EDA and feature analysis.
- Correlation does not imply causation.
- A correlation close to zero only indicates little or no linear relationship for Pearson correlation.
