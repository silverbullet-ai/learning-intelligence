# Standard Deviation

This notebook demonstrates:

- Population Standard Deviation
- Sample Standard Deviation
- Relationship between Variance and Standard Deviation
- Visual interpretation of spread

```python
import statistics
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data = [1, 2, 3, 4, 5]

df = pd.DataFrame({"Values": data})

df
```

```python
mean = statistics.mean(data)

print("Mean:", mean)
```

```python
population_variance = statistics.pvariance(data)
sample_variance = statistics.variance(data)

print("Population Variance :", population_variance)
print("Sample Variance     :", sample_variance)
```

```python
population_std = statistics.pstdev(data)
sample_std = statistics.stdev(data)

print("Population Standard Deviation :", population_std)
print("Sample Standard Deviation     :", sample_std)
```

```python
print("√Population Variance =", np.sqrt(population_variance))
print("Population Std Dev   =", population_std)
```

```python
comparison = pd.DataFrame({
    "Metric": [
        "Mean",
        "Population Variance",
        "Sample Variance",
        "Population Std Dev",
        "Sample Std Dev"
    ],
    "Value": [
        mean,
        population_variance,
        sample_variance,
        population_std,
        sample_std
    ]
})

comparison
```

```python
plt.figure(figsize=(8,4))

plt.scatter(data, [1]*len(data), s=120)

plt.axvline(mean, color="red", linestyle="--", label="Mean")

plt.title("Observations Around the Mean")

plt.yticks([])

plt.legend()

plt.show()
```

```python
print("Mean :", mean)
print("Standard Deviation :", round(population_std, 2))

for value in data:
    distance = value - mean
    print(f"{value} -> Distance from Mean = {distance}")
```

## Summary

- Standard Deviation is the square root of Variance.
- It measures the spread of observations around the mean.
- Population Standard Deviation uses **σ**.
- Sample Standard Deviation uses **s**.
- Standard Deviation is easier to interpret because it uses the same units as the original data.
