```python
import pandas as pd
import matplotlib.pyplot as plt
import statistics
```

## Creating a Sample Dataset

We'll use a simple dataset containing the ages of a small group of people.

```python
ages = [24, 27, 14, 13, 28, 29, 31, 32]

df = pd.DataFrame({
    "Age": ages
})

df
```

## Descriptive Statistics

Pandas provides a convenient method called `describe()`
that summarizes numerical data.

```python
df.describe()
```

## Mean

```python
print("Mean:", statistics.mean(ages))
```

## Median

```python
print("Median:", statistics.median(ages))
```

## Mode

```python
try:
    print("Mode:", statistics.mode(ages))
except statistics.StatisticsError:
    print("No unique mode found.")
```

## Variance

```python
print("Variance:", statistics.variance(ages))
```

# Histogram

A histogram shows how frequently values occur within different intervals.

```python
plt.figure(figsize=(7,4))

plt.hist(
    ages,
    bins=5,
    edgecolor="black"
)

plt.title("Age Distribution")

plt.xlabel("Age")

plt.ylabel("Frequency")

plt.grid(alpha=0.3)

plt.show()
```

The histogram helps us observe:

- Distribution of values
- Concentration of data
- Possible skewness
- Frequency

# Box Plot

A box plot summarizes the spread of data and helps identify potential outliers.

```python
plt.figure(figsize=(6,2))

plt.boxplot(
    ages,
    vert=False
)

plt.title("Age Box Plot")

plt.show()
```

The box plot shows:

- Minimum
- First Quartile (Q1)
- Median
- Third Quartile (Q3)
- Maximum

It is commonly used for detecting outliers.
