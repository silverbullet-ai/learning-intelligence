```python
import pandas as pd
import matplotlib.pyplot as plt
import statistics
```

```python
data = [1,3,4,5]

df = pd.DataFrame({
    "Values": data
})

df
```

```python
statistics.mean(data)
```

```python
statistics.median(data)
```

```python
statistics.mode([4,3,2,1,1,4,4,5,2,100])
```

```python
without_outlier = [1,3,4,5]

with_outlier = [1,3,4,5,100]

print("Mean without outlier :", statistics.mean(without_outlier))
print("Mean with outlier    :", statistics.mean(with_outlier))

print()

print("Median without outlier :", statistics.median(without_outlier))
print("Median with outlier    :", statistics.median(with_outlier))
```

```python
plt.hist(
    with_outlier,
    bins=5,
    edgecolor="black"
)

plt.title("Histogram with Outlier")

plt.show()
```

```python
plt.boxplot(
    with_outlier,
    vert=False
)

plt.title("Box Plot Showing Outlier")

plt.show()
```

```python
mode_data = [4,3,2,1,1,4,4,5,2,100]

pd.Series(mode_data).value_counts()
```

## Summary

✔ Mean changes significantly due to outliers.

✔ Median remains stable.

✔ Mode depends only on frequency.

✔ Choose the measure based on the characteristics of the data.
