# Variables in Statistics

This notebook demonstrates:

- Quantitative Variables
- Discrete Variables
- Continuous Variables
- Qualitative Variables

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.DataFrame({
    "Age": [21, 24, 22, 25, 23],
    "Height": [170.5, 168.2, 175.4, 180.0, 172.3],
    "Students": [30, 35, 28, 40, 33],
    "Gender": ["Male", "Female", "Male", "Female", "Male"]
})

data
```

```python
variable_types = pd.DataFrame({
    "Variable": ["Age", "Height", "Students", "Gender"],
    "Type": [
        "Continuous",
        "Continuous",
        "Discrete",
        "Qualitative"
    ]
})

variable_types
```

```python
variable_types = pd.DataFrame({
    "Variable": ["Age", "Height", "Students", "Gender"],
    "Type": [
        "Continuous",
        "Continuous",
        "Discrete",
        "Qualitative"
    ]
})

variable_types
```

```python
plt.hist(data["Height"], bins=5)

plt.title("Continuous Variable: Height")

plt.xlabel("Height (cm)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.bar(
    data.index,
    data["Students"]
)

plt.title("Discrete Variable: Number of Students")

plt.xlabel("Observation")
plt.ylabel("Students")

plt.show()
```

```python
data["Gender"].value_counts().plot(kind="bar")

plt.title("Qualitative Variable: Gender")

plt.xlabel("Gender")
plt.ylabel("Count")

plt.show()
```

## Summary

- Quantitative variables are numerical.
- Discrete variables are countable.
- Continuous variables are measurable.
- Qualitative variables represent categories.
