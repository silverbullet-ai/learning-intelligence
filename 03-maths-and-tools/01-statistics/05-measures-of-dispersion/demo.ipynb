# Measures of Dispersion

This notebook demonstrates:

- Mean
- Variance
- Population Variance
- Sample Variance
- Low vs High Variance

```python
import statistics
import pandas as pd
import matplotlib.pyplot as plt
```

```python
distribution1 = [2,2,4,4]

distribution2 = [1,1,5,5]
```

```python
print("Mean Distribution 1 :", statistics.mean(distribution1))
print("Mean Distribution 2 :", statistics.mean(distribution2))
```

```python
print("Population Variance 1 :", statistics.pvariance(distribution1))
print("Population Variance 2 :", statistics.pvariance(distribution2))
```

```python
print("Sample Variance 1 :", statistics.variance(distribution1))
print("Sample Variance 2 :", statistics.variance(distribution2))
```

```python
fig, ax = plt.subplots(figsize=(8,2))

ax.scatter(
    distribution1,
    [1]*len(distribution1),
    s=120,
    label="Distribution 1"
)

ax.scatter(
    distribution2,
    [0]*len(distribution2),
    s=120,
    label="Distribution 2"
)

ax.set_yticks([0,1])
ax.set_yticklabels(["Distribution 2","Distribution 1"])

plt.title("Comparing Two Distributions")

plt.legend()

plt.show()
```

```python
plt.figure(figsize=(7,4))

plt.hist(
    distribution1,
    alpha=0.6,
    bins=4,
    label="Distribution 1"
)

plt.hist(
    distribution2,
    alpha=0.6,
    bins=4,
    label="Distribution 2"
)

plt.legend()

plt.title("Histogram Comparison")

plt.show()
```

```python
comparison = pd.DataFrame({
    "Distribution":["D1","D2"],
    "Mean":[
        statistics.mean(distribution1),
        statistics.mean(distribution2)
    ],
    "Population Variance":[
        statistics.pvariance(distribution1),
        statistics.pvariance(distribution2)
    ]
})

comparison
```

# Summary

- Both datasets have the same mean.
- Distribution 2 has a larger variance.
- Variance measures how spread out observations are around the mean.
- Larger variance indicates greater variability.
