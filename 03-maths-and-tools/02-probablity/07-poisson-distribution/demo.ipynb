# Poisson Distribution

A practical demonstration of the **Poisson Distribution** using Python.

We will calculate probabilities, verify the mean/variance properties, visualize the PMF, and compare different values of `λ`.

## 1. Import Libraries

We use NumPy for numerical calculations, SciPy for the Poisson distribution, and Matplotlib for visualization.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson
```

## 2. Poisson PMF

The probability mass function is:

$$P(X=k)=\\frac{e^{-\\lambda}\\lambda^k}{k!}$$

Let's calculate the probability of exactly 5 customers arriving when the average is 3 customers per hour.

```python
lam = 3
k = 5

probability = poisson.pmf(k, mu=lam)

print(f"P(X = {k}) = {probability:.4f}")
print(f"Percentage = {probability * 100:.2f}%")
```

## 3. Mean, Variance and Standard Deviation

For a Poisson Distribution:

$$Mean = \\lambda$$

$$Variance = \\lambda$$

$$Standard\ Deviation = \\sqrt{\\lambda}$$

```python
mean = poisson.mean(mu=lam)
variance = poisson.var(mu=lam)
std_dev = poisson.std(mu=lam)

print(f"λ = {lam}")
print(f"Mean = {mean:.2f}")
print(f"Variance = {variance:.2f}")
print(f"Standard Deviation = {std_dev:.2f}")
```

## 4. Different Probability Questions

For `λ = 3`:

- Exactly 5 → `P(X = 5)`
- At most 3 → `P(X ≤ 3)`
- More than 3 → `P(X > 3)`
- Between 4 and 5 → `P(4 ≤ X ≤ 5)`

```python
exactly_5 = poisson.pmf(5, mu=lam)
at_most_3 = poisson.cdf(3, mu=lam)
more_than_3 = poisson.sf(3, mu=lam)
between_4_and_5 = poisson.cdf(5, mu=lam) - poisson.cdf(3, mu=lam)

print(f"P(X = 5)       = {exactly_5:.4f}")
print(f"P(X <= 3)      = {at_most_3:.4f}")
print(f"P(X > 3)       = {more_than_3:.4f}")
print(f"P(4 <= X <= 5) = {between_4_and_5:.4f}")
```

## 5. Visualizing the Poisson PMF

Because Poisson is a **discrete distribution**, its PMF is represented using separate bars.

```python
x = np.arange(0, 16)
pmf = poisson.pmf(x, mu=lam)

plt.figure(figsize=(9, 5))
plt.bar(x, pmf)
plt.xlabel("Number of Events (X)")
plt.ylabel("P(X = x)")
plt.title("Poisson Distribution (λ = 3)")
plt.xticks(x)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## 6. Effect of λ

Let's compare Poisson distributions with different values of `λ`.

As `λ` increases, the distribution shifts to the right and becomes more symmetric.

```python
lambdas = [1, 4, 10]
x = np.arange(0, 26)

plt.figure(figsize=(10, 6))

for value in lambdas:
    plt.plot(x, poisson.pmf(x, mu=value), marker="o", label=f"λ = {value}")

plt.xlabel("Number of Events (X)")
plt.ylabel("P(X = x)")
plt.title("Effect of λ on the Poisson Distribution")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## 7. Rate and Time Interval

Suppose a call center receives an average of **3 calls per hour**.

For a 2-hour interval:

$$\\lambda_t = 3 \\times 2 = 6$$

So the number of calls in two hours follows `Poisson(6)`.

```python
rate_per_hour = 3
time_hours = 2
effective_lambda = rate_per_hour * time_hours

probability_8_calls = poisson.pmf(8, mu=effective_lambda)

print(f"Effective λ = {effective_lambda}")
print(f"P(X = 8) = {probability_8_calls:.4f}")
print(f"Percentage = {probability_8_calls * 100:.2f}%")
```

## 8. Quick Practice

## Key Takeaways

- Poisson is a **discrete probability distribution**.
- It models the **number of events in a fixed interval**.
- `λ` is the average number of events per interval.
- Mean = Variance = `λ`.
- Standard deviation = `√λ`.
- For a rate over time `t`, use `λt`.
- Use PMF for exact counts and CDF/complement methods for cumulative questions.
