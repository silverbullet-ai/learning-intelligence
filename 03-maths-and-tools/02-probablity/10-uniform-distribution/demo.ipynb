# Uniform Distribution — Demo

This notebook demonstrates both forms of the **Uniform Distribution**:

- Continuous Uniform Distribution
- Discrete Uniform Distribution

We will calculate probabilities, visualize the distributions, and verify their statistical properties.

## 1. Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import uniform
```

# Part A — Continuous Uniform Distribution

## 2. Define a Continuous Uniform Distribution

Let:

$$
X \sim U(10,40)
$$

Therefore:

- $a = 10$
- $b = 40$

```python
a = 10
b = 40

print("Lower bound (a):", a)
print("Upper bound (b):", b)
print("Interval length:", b - a)
```

## 3. PDF

For a Continuous Uniform Distribution:

$$
f(x)=\frac{1}{b-a}
$$

for $a \leq x \leq b$.

```python
density = 1 / (b - a)

print(f"Constant probability density = {density:.4f}")
```

## 4. Visualize the Continuous Uniform PDF

```python
x = np.linspace(5, 45, 1000)
pdf = uniform.pdf(x, loc=a, scale=b-a)

plt.figure(figsize=(10, 5))
plt.plot(x, pdf)
plt.xlabel("X")
plt.ylabel("Probability Density")
plt.title("Continuous Uniform Distribution: U(10, 40)")
plt.grid(True)
plt.show()
```

## 5. Probability Between Two Values

Find:

$$
P(15 \leq X \leq 30)
$$

For a Continuous Uniform Distribution:

$$
P(c \leq X \leq d)=\frac{d-c}{b-a}
$$

```python
c = 15
d = 30

probability = (d - c) / (b - a)

print(f"P({c} ≤ X ≤ {d}) = {probability:.4f}")
print(f"Percentage = {probability * 100:.2f}%")
```

## 6. Probability of X ≥ 20

Find:

$$
P(X \geq 20)
$$

The interval from 20 to 40 has length 20.

```python
lower = 20

probability = (b - lower) / (b - a)

print(f"P(X ≥ {lower}) = {probability:.4f}")
print(f"Percentage = {probability * 100:.2f}%")
```

The exact result is:

$$
\frac{20}{30}=\frac{2}{3}\approx0.6667
$$

So the probability is approximately **66.67%**.

## 7. Mean, Variance and Standard Deviation

For a Continuous Uniform Distribution:

$$
\mu=\frac{a+b}{2}
$$

$$
\sigma^2=\frac{(b-a)^2}{12}
$$

$$
\sigma=\frac{b-a}{\sqrt{12}}
$$

```python
mean = (a + b) / 2
variance = (b - a) ** 2 / 12
std_dev = np.sqrt(variance)

print(f"Mean = {mean:.4f}")
print(f"Variance = {variance:.4f}")
print(f"Standard Deviation = {std_dev:.4f}")
```

## 8. CDF

The CDF represents:

$$
F(x)=P(X\leq x)
$$

For the Continuous Uniform Distribution:

$$
F(x)=\frac{x-a}{b-a}
$$

inside the interval $[a,b]$.

```python
x_values = np.array([10, 15, 20, 25, 30, 35, 40])
cdf_values = uniform.cdf(x_values, loc=a, scale=b-a)

for x_value, cdf_value in zip(x_values, cdf_values):
    print(f"F({x_value}) = {cdf_value:.4f}")
```

## 9. Generate Random Samples

A Uniform Distribution can also be used to generate random values within a fixed interval.

```python
np.random.seed(42)

samples = np.random.uniform(a, b, 10000)

print("First 10 samples:")
print(np.round(samples[:10], 2))
```

```python
plt.figure(figsize=(10, 5))
plt.hist(samples, bins=30, density=True)
plt.xlabel("X")
plt.ylabel("Density")
plt.title("Samples from U(10, 40)")
plt.grid(True)
plt.show()
```

# Part B — Discrete Uniform Distribution

## 10. Fair Die

A fair six-sided die has six equally likely outcomes:

$$
\{1,2,3,4,5,6\}
$$

Each outcome has probability:

$$
P(X=x)=\frac{1}{6}
$$

```python
outcomes = np.arange(1, 7)
probability = 1 / len(outcomes)

for outcome in outcomes:
    print(f"P(X = {outcome}) = {probability:.4f}")
```

## 11. PMF Visualization

```python
probabilities = np.full(len(outcomes), 1 / len(outcomes))

plt.figure(figsize=(8, 5))
plt.bar(outcomes, probabilities)
plt.xlabel("Outcome")
plt.ylabel("P(X = x)")
plt.title("Discrete Uniform Distribution — Fair Die")
plt.xticks(outcomes)
plt.grid(axis="y")
plt.show()
```

## 12. Mean of a Fair Die

For consecutive integer outcomes from $a$ to $b$:

$$
\mu=\frac{a+b}{2}
$$

```python
a_discrete = 1
b_discrete = 6

mean_die = (a_discrete + b_discrete) / 2

print(f"Mean of fair die = {mean_die}")
```

Notice that the expected value is **3.5**, even though 3.5 is not a possible outcome of a single die roll.

## 13. Variance of a Discrete Uniform Distribution

For consecutive integer outcomes from $a$ to $b$:

$$
n=b-a+1
$$

and:

$$
\sigma^2=\frac{n^2-1}{12}
$$

```python
n = b_discrete - a_discrete + 1
variance_die = (n ** 2 - 1) / 12
std_die = np.sqrt(variance_die)

print(f"Number of outcomes = {n}")
print(f"Variance = {variance_die:.4f}")
print(f"Standard Deviation = {std_die:.4f}")
```

## 14. Fair Coin as a Discrete Uniform Distribution

A fair coin has two equally likely outcomes:

$$
P(Heads)=\frac{1}{2}
$$

$$
P(Tails)=\frac{1}{2}
$$

```python
coin_outcomes = ["Heads", "Tails"]
coin_probability = 1 / len(coin_outcomes)

for outcome in coin_outcomes:
    print(f"P({outcome}) = {coin_probability:.2f}")
```

## Key Takeaways

- Continuous Uniform Distribution uses a **PDF**.
- Discrete Uniform Distribution uses a **PMF**.
- Continuous Uniform Distribution has constant probability density over its interval.
- A continuous variable has probability 0 at any exact point.
- Discrete Uniform Distribution assigns equal probability to every possible outcome.
- Continuous Uniform Mean = `(a + b) / 2`.
- Continuous Uniform Variance = `(b - a)^2 / 12`.
- For `n` equally likely discrete outcomes, each has probability `1/n`.
