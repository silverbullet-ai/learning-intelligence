# Bernoulli Distribution

This notebook demonstrates the Bernoulli Distribution using Python.

We will:

- Define a Bernoulli random variable
- Calculate its PMF
- Visualize the distribution
- Calculate mean and variance
- Use SciPy's Bernoulli distribution
- Simulate Bernoulli trials
- Compare theoretical and simulated results

## Bernoulli Random Variable

A Bernoulli random variable has exactly two possible outcomes:

$$
X\in\{0,1\}
$$

We will use:

- \(1\) → Success
- \(0\) → Failure

Suppose the probability of success is:

$$
p=0.6
$$

Therefore:

$$
P(X=1)=0.6
$$

and:

$$
P(X=0)=1-0.6=0.4
$$

```python
import numpy as np
import matplotlib.pyplot as plt

p = 0.6

x = np.array([0, 1])
probabilities = np.array([1 - p, p])

print("Possible outcomes:", x)
print("Probabilities:", probabilities)
print("Total probability:", probabilities.sum())
```

```python
plt.bar(x, probabilities)

plt.xticks([0, 1], ["Failure (0)", "Success (1)"])
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.title("Bernoulli Distribution")

plt.ylim(0, 1)

plt.show()
```

## PMF

The Bernoulli PMF is:

$$
P(X=k)=p^k(1-p)^{1-k}
$$

where:

$$
k\in\{0,1\}
$$

Let's calculate the PMF directly using the formula.

```python
def bernoulli_pmf(k, p):
    return (p ** k) * ((1 - p) ** (1 - k))


for k in [0, 1]:
    print(f"P(X={k}) =", bernoulli_pmf(k, p))
```

## Mean

For a Bernoulli Distribution:

$$
E[X]=p
$$

For \(p=0.6\):

$$
E[X]=0.6
$$

```python
mean = p

print("Theoretical mean:", mean)
```

## Variance and Standard Deviation

The variance of a Bernoulli Distribution is:

$$
\operatorname{Var}(X)=p(1-p)
$$

The standard deviation is:

$$
\sigma=\sqrt{p(1-p)}
$$

```python
variance = p * (1 - p)
std = np.sqrt(variance)

print("Variance:", variance)
print("Standard deviation:", std)
```

## Using SciPy

SciPy provides a ready-made Bernoulli distribution.

We can use it to calculate:

- PMF
- Mean
- Variance
- Random samples

```python
from scipy.stats import bernoulli

distribution = bernoulli(p)

print("P(X=0):", distribution.pmf(0))
print("P(X=1):", distribution.pmf(1))
print("Mean:", distribution.mean())
print("Variance:", distribution.var())
```

## Simulating Bernoulli Trials

Theoretical probability tells us what we expect mathematically.

Simulation allows us to observe what happens experimentally.

Let's simulate 10,000 Bernoulli trials where:

$$
p=0.6
$$

```python
np.random.seed(42)

n_trials = 10_000

samples = bernoulli.rvs(p, size=n_trials)

successes = samples.sum()
failures = n_trials - successes

print("Total trials:", n_trials)
print("Successes:", successes)
print("Failures:", failures)

print("Observed success probability:", successes / n_trials)
```

## Theoretical vs Observed Probability

The theoretical probability of success is:

$$
P(X=1)=0.6
$$

The simulated probability should be close to \(0.6\), although it will not normally be exactly \(0.6\).

As the number of trials increases, the observed probability generally gets closer to the theoretical probability.

```python
theoretical_probability = p
observed_probability = samples.mean()

print(f"Theoretical probability: {theoretical_probability:.4f}")
print(f"Observed probability:    {observed_probability:.4f}")
```

## Why Bernoulli Matters

Bernoulli trials are the building blocks of several important statistical concepts.

For example:

$$
\text{Bernoulli trials}
\longrightarrow
\text{Binomial Distribution}
$$

A Bernoulli trial models one binary outcome.

A Binomial Distribution counts the number of successes across multiple independent Bernoulli trials.

This relationship will be explored in the next topic.
