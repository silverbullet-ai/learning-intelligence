# Normal (Gaussian) Distribution — Demo

This notebook demonstrates the key concepts of the **Normal Distribution** using Python.

Topics covered:
- Normal PDF
- Mean and standard deviation
- Effect of mean and variance
- Probability using the Normal distribution
- Empirical Rule (68–95–99.7)
- Standard Normal Distribution
- Z-scores
- Visualization

## 1. Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

## 2. Define a Normal Distribution

A Normal Distribution is represented as:

$$
X \sim N(\mu, \sigma^2)
$$

Let's use:

- Mean = 100
- Standard deviation = 10

```python
mu = 100
sigma = 10

print("Mean:", mu)
print("Standard Deviation:", sigma)
print("Variance:", sigma ** 2)
```

## 3. Generate the Normal Distribution Curve

```python
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 1000)
y = norm.pdf(x, mu, sigma)

plt.figure(figsize=(10, 5))
plt.plot(x, y)
plt.title("Normal Distribution")
plt.xlabel("X")
plt.ylabel("Probability Density")
plt.grid(True)
plt.show()
```

## 4. Probability Density Function

The PDF of a Normal Distribution is:

$$
f(x) = \frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$

Python's `scipy.stats.norm.pdf()` can calculate the density directly.

```python
x_value = 100
density = norm.pdf(x_value, mu, sigma)

print(f"PDF at x = {x_value}: {density:.6f}")
```

## 5. Probability Within One Standard Deviation

According to the empirical rule, approximately 68% of observations lie within:

$$
\mu - \sigma \leq X \leq \mu + \sigma
$$

```python
lower = mu - sigma
upper = mu + sigma

probability = norm.cdf(upper, mu, sigma) - norm.cdf(lower, mu, sigma)

print(f"Range: {lower} to {upper}")
print(f"Probability: {probability:.4f}")
print(f"Percentage: {probability * 100:.2f}%")
```

## 6. Empirical Rule: 68–95–99.7

Let's calculate the probability within 1, 2, and 3 standard deviations.

```python
for k in [1, 2, 3]:
    lower = mu - k * sigma
    upper = mu + k * sigma
    probability = norm.cdf(upper, mu, sigma) - norm.cdf(lower, mu, sigma)
    print(f"Within ±{k}σ: {probability * 100:.2f}%")
```

Expected results:

- ±1σ → approximately 68%
- ±2σ → approximately 95%
- ±3σ → approximately 99.7%

## 7. Effect of the Mean

The mean controls the **location** of the Normal Distribution.

Below, the standard deviation remains constant while the mean changes.

```python
x = np.linspace(-10, 20, 1000)

means = [0, 5, 10]
sigma = 2

plt.figure(figsize=(10, 5))

for mean in means:
    plt.plot(x, norm.pdf(x, mean, sigma), label=f"μ = {mean}")

plt.title("Effect of Mean on Normal Distribution")
plt.xlabel("X")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(True)
plt.show()
```

## 8. Effect of Variance

The variance controls the **spread** of the Normal Distribution.

Here the mean remains constant while the standard deviation changes.

```python
mu = 0
x = np.linspace(-10, 10, 1000)
standard_deviations = [1, 2, 3]

plt.figure(figsize=(10, 5))

for sigma in standard_deviations:
    plt.plot(x, norm.pdf(x, mu, sigma), label=f"σ = {sigma}")

plt.title("Effect of Standard Deviation on Normal Distribution")
plt.xlabel("X")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(True)
plt.show()
```

## 9. Standard Normal Distribution

The Standard Normal Distribution has:

$$
\mu = 0
$$

$$
\sigma = 1
$$

Therefore:

$$
Z \sim N(0,1)
$$

```python
x = np.linspace(-4, 4, 1000)
y = norm.pdf(x, 0, 1)

plt.figure(figsize=(10, 5))
plt.plot(x, y)
plt.title("Standard Normal Distribution")
plt.xlabel("Z-score")
plt.ylabel("Probability Density")
plt.grid(True)
plt.show()
```

## 10. Z-Score

The Z-score tells us how many standard deviations an observation is away from the mean.

$$
Z = \frac{X - \mu}{\sigma}
$$

```python
x_value = 120
mu = 100
sigma = 10

z = (x_value - mu) / sigma

print("X:", x_value)
print("Mean:", mu)
print("Standard Deviation:", sigma)
print("Z-score:", z)
```

### Interpretation

For the example above:

$$
Z = 2
$$

This means the observation is **2 standard deviations above the mean**.

## 11. Probability Using a Z-Score

Suppose:

- Mean = 100
- Standard deviation = 10

Find the probability that $X \leq 120$.

```python
mu = 100
sigma = 10
x_value = 120

z = (x_value - mu) / sigma
probability = norm.cdf(x_value, mu, sigma)

print(f"Z-score: {z:.2f}")
print(f"P(X ≤ {x_value}) = {probability:.4f}")
print(f"Percentage = {probability * 100:.2f}%")
```

## 12. Probability Between Two Values

Find:

$$
P(90 \leq X \leq 110)
$$

for a Normal Distribution with mean 100 and standard deviation 10.

```python
mu = 100
sigma = 10

lower = 90
upper = 110

probability = norm.cdf(upper, mu, sigma) - norm.cdf(lower, mu, sigma)

print(f"P({lower} ≤ X ≤ {upper}) = {probability:.4f}")
print(f"Percentage = {probability * 100:.2f}%")
```

## 13. Key Takeaways

- Normal Distribution is **continuous**.
- It uses a **PDF**.
- Its curve is **bell-shaped and symmetric**.
- Mean = Median = Mode.
- Mean controls the location of the distribution.
- Variance controls the spread.
- Standard deviation is the square root of variance.
- Approximately 68% of observations lie within ±1σ.
- Approximately 95% lie within ±2σ.
- Approximately 99.7% lie within ±3σ.
- Standard Normal Distribution has μ = 0 and σ = 1.
- Z-score is calculated using:

$$
Z = \frac{X-\mu}{\sigma}
$$
