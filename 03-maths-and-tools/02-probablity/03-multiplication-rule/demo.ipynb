# Multiplication Rule

The Multiplication Rule calculates the probability that **two events occur together**.

For two events \(A\) and \(B\):

$$
P(A \cap B) = P(A)P(B \mid A)
$$

For independent events:

$$
P(A \cap B) = P(A)P(B)
$$

We will demonstrate both independent and dependent events using coin, dice, and card examples.

## Independent Events: Coin Toss

A fair coin is tossed twice.

We want:

- Head on the first toss
- Tail on the second toss

The two tosses are independent because the first toss does not affect the second.

$$
P(H \cap T) = P(H)P(T)
$$

```python
p_head = 1 / 2
p_tail = 1 / 2

p_head_and_tail = p_head * p_tail

print(f"P(Head) = {p_head}")
print(f"P(Tail) = {p_tail}")
print(f"P(Head and Tail) = {p_head_and_tail}")
print(f"Percentage = {p_head_and_tail * 100:.2f}%")
```

## Independent Events: Dice Rolls

A fair six-sided die is rolled twice.

Find the probability of rolling:

- \(1\) on the first roll
- \(2\) on the second roll

Since the first roll does not affect the second roll, the events are independent.

```python
p_first_1 = 1 / 6
p_second_2 = 1 / 6

p_both = p_first_1 * p_second_2

print(f"P(1 on first roll) = {p_first_1:.4f}")
print(f"P(2 on second roll) = {p_second_2:.4f}")
print(f"P(1 then 2) = {p_both:.4f}")
```

## Dependent Events: Drawing Cards Without Replacement

Consider a standard deck of 52 cards.

We draw:

1. A King
2. A Queen

The first card is **not replaced**.

Therefore, the second probability depends on the first draw.

$$
P(K \cap Q) = P(K)P(Q \mid K)
$$

```python
total_cards = 52
kings = 4
queens = 4

p_king = kings / total_cards
p_queen_given_king = queens / (total_cards - 1)

p_king_and_queen = p_king * p_queen_given_king

print(f"P(King) = {p_king:.6f}")
print(f"P(Queen | King) = {p_queen_given_king:.6f}")
print(f"P(King and Queen) = {p_king_and_queen:.6f}")
print(f"Percentage = {p_king_and_queen * 100:.3f}%")
```

## Conditional Probability

The conditional probability of \(B\) given \(A\) is:

$$
P(B \mid A)
=
\frac{P(A \cap B)}{P(A)}
$$

For the card example:

$$
P(Q \mid K)=\frac{4}{51}
$$

because one King has already been removed.

```python
p_king_and_queen = 4 / 52 * 4 / 51
p_king = 4 / 52

p_queen_given_king = p_king_and_queen / p_king

print(f"P(King and Queen) = {p_king_and_queen:.6f}")
print(f"P(King) = {p_king:.6f}")
print(f"P(Queen | King) = {p_queen_given_king:.6f}")
```

## With Replacement vs. Without Replacement

Replacing an item restores the original sample space.

Without replacement, the sample space changes.

This is why:

- **With replacement** → events can be independent.
- **Without replacement** → events are generally dependent.

Let's compare drawing two Kings with and without replacement.

```python
# Two Kings drawn with replacement

p_king_with_replacement = (4 / 52) * (4 / 52)

# Two Kings drawn without replacement

p_king_without_replacement = (4 / 52) * (3 / 51)

print("Two Kings with replacement:")
print(f"{p_king_with_replacement:.6f}")

print("\nTwo Kings without replacement:")
print(f"{p_king_without_replacement:.6f}")
```

## Comparing the Two Cases

With replacement:

$$
P(K \cap K)
=
\frac{4}{52}\times\frac{4}{52}
$$

Without replacement:

$$
P(K \cap K)
=
\frac{4}{52}\times\frac{3}{51}
$$

The second probability changes because the first King was not replaced.

```python
with_replacement = (4 / 52) * (4 / 52)
without_replacement = (4 / 52) * (3 / 51)

print(f"With replacement:    {with_replacement:.6f}")
print(f"Without replacement: {without_replacement:.6f}")
```

## Quick Comparison

| Concept | Formula |
|---|---|
| Independent Events | \(P(A \cap B)=P(A)P(B)\) |
| Dependent Events | \(P(A \cap B)=P(A)P(B\mid A)\) |
| Conditional Probability | \(P(B\mid A)=\frac{P(A\cap B)}{P(A)}\) |

The key question to ask is:

> **Does the first event change the probability of the second event?**
