# Addition Rule

This notebook demonstrates the Addition Rule using simple dice examples.

We will explore:

- Union of events
- Intersection of events
- Mutually exclusive events
- Non-mutually exclusive events
- General Addition Rule

## Union and Intersection

For two events \(A\) and \(B\):

- Union:

$$
A \cup B
$$

means A **or** B **or both**.

- Intersection:

$$
A \cap B
$$

means A **and** B.

```python
sample_space = {1, 2, 3, 4, 5, 6}

A = {2, 4, 6}       # Even numbers
B = {4, 5, 6}       # Numbers greater than 3

union = A | B
intersection = A & B

print("A:", A)
print("B:", B)
print("A ∪ B:", union)
print("A ∩ B:", intersection)
```

## General Addition Rule

For any two events:

$$
P(A \cup B)
=
P(A)+P(B)-P(A\cap B)
$$

We subtract the intersection because those outcomes are counted twice when adding \(P(A)\) and \(P(B)\).

```python
p_A = len(A) / len(sample_space)
p_B = len(B) / len(sample_space)
p_intersection = len(intersection) / len(sample_space)

p_union = p_A + p_B - p_intersection

print(f"P(A) = {p_A:.4f}")
print(f"P(B) = {p_B:.4f}")
print(f"P(A ∩ B) = {p_intersection:.4f}")
print(f"P(A ∪ B) = {p_union:.4f}")
```

## Mutually Exclusive Events

Events are mutually exclusive when they cannot occur together.

Therefore:

$$
A\cap B=\varnothing
$$

and:

$$
P(A\cap B)=0
$$

The Addition Rule becomes:

$$
P(A\cup B)=P(A)+P(B)
$$

```python
A = {2}
B = {5}

intersection = A & B
union = A | B

p_A = len(A) / len(sample_space)
p_B = len(B) / len(sample_space)
p_union = len(union) / len(sample_space)

print("A ∩ B:", intersection)
print(f"P(A) = {p_A:.4f}")
print(f"P(B) = {p_B:.4f}")
print(f"P(A ∪ B) = {p_union:.4f}")
print(f"P(A) + P(B) = {p_A + p_B:.4f}")
```

## Non-Mutually Exclusive Events

Events are non-mutually exclusive when they can occur together.

For example:

- \(A\): rolling an even number
- \(B\): rolling a number greater than \(3\)

These events overlap at \(4\) and \(6\).

Therefore:

$$
A\cap B=\{4,6\}
$$

```python
A = {2, 4, 6}
B = {4, 5, 6}

intersection = A & B
union = A | B

p_A = len(A) / len(sample_space)
p_B = len(B) / len(sample_space)
p_intersection = len(intersection) / len(sample_space)

p_union = p_A + p_B - p_intersection

print("A ∩ B:", intersection)
print("A ∪ B:", union)
print(f"P(A) = {p_A:.4f}")
print(f"P(B) = {p_B:.4f}")
print(f"P(A ∩ B) = {p_intersection:.4f}")
print(f"P(A ∪ B) = {p_union:.4f}")
```

## Verify Using Direct Counting

Because we know the complete sample space, we can directly count the outcomes in the union.

The Addition Rule should produce the same result.

```python
import math

direct_probability = len(union) / len(sample_space)

print(f"Direct probability = {direct_probability:.4f}")
print(f"Addition Rule      = {p_union:.4f}")

assert math.isclose(direct_probability, p_union)
print("Both methods agree.")
```
