# Probability Basics

This notebook demonstrates the fundamental concepts of probability using simple Python examples.

We will work with:

- Sample spaces
- Events
- Basic probability
- Probability ranges
- Complements

## Basic Probability

For equally likely outcomes:

$$
P(A) = \frac{\text{Number of favorable outcomes}}
{\text{Total number of possible outcomes}}
$$

For a fair six-sided die, the probability of rolling a 4 is:

$$
P(4) = \frac{1}{6}
$$

```python
favorable_outcomes = 1
total_outcomes = 6

probability = favorable_outcomes / total_outcomes

print(f"P(rolling a 4) = {probability:.4f}")
print(f"Percentage = {probability * 100:.2f}%")
```

## Sample Space and Events

Consider a six-sided die.

The sample space is:

$$
S = \{1,2,3,4,5,6\}
$$

The event of rolling an even number is:

$$
A = \{2,4,6\}
$$

## Mutually Exclusive Events

Two events are mutually exclusive when they cannot occur at the same time.

Example:

- Getting Head
- Getting Tail

in a single coin toss.

```python
sample_space = {1, 2, 3, 4, 5, 6}
even_event = {2, 4, 6}

probability_even = len(even_event) / len(sample_space)

print("Sample space:", sample_space)
print("Even-number event:", even_event)
print(f"P(Even) = {probability_even}")
```

## Complement of an Event

The complement of an event represents the event not occurring.

$$
P(A^c) = 1 - P(A)
$$

If the probability of rolling an even number is \(0.5\), then:

$$
P(\text{Odd}) = 1 - 0.5 = 0.5
$$

```python
probability_even = 0.5
probability_odd = 1 - probability_even

print(f"P(Even) = {probability_even}")
print(f"P(Odd) = {probability_odd}")
print(f"Sum = {probability_even + probability_odd}")
```

```python
# Simple simulation of rolling a fair die

import random

rolls = [random.randint(1, 6) for _ in range(10_000)]

even_count = sum(roll % 2 == 0 for roll in rolls)
experimental_probability = even_count / len(rolls)

print(f"Experimental P(Even) ≈ {experimental_probability:.4f}")
print("Theoretical P(Even) =", 3 / 6)
```
