# Probability

Probability measures the likelihood of an event occurring.

In this notebook, we will explore:

- Basic probability
- Coin toss
- Dice rolls
- Mutually exclusive events
- Non-mutually exclusive events
- Addition rules
- A deck-of-cards example

```python
# Basic probability

favorable_outcomes = 1
total_outcomes = 2

probability = favorable_outcomes / total_outcomes

print(f"Probability = {probability}")
print(f"Percentage = {probability * 100}%")
```

```python
# Probability of Head and Tail

p_head = 1 / 2
p_tail = 1 / 2

print(f"P(Head) = {p_head}")
print(f"P(Tail) = {p_tail}")
print(f"P(Head or Tail) = {p_head + p_tail}")
```

```python
# Probability of rolling a specific number

total_outcomes = 6
favorable_outcomes = 1

p_one = favorable_outcomes / total_outcomes
p_five = favorable_outcomes / total_outcomes

print(f"P(1) = {p_one}")
print(f"P(5) = {p_five}")
print(f"P(1 or 5) = {p_one + p_five}")
```

## Mutually Exclusive Events

Two events are mutually exclusive when they cannot occur at the same time.

Example:

- Getting Head
- Getting Tail

in a single coin toss.

```python
# Addition rule for mutually exclusive events

p_head = 1 / 2
p_tail = 1 / 2

p_head_or_tail = p_head + p_tail

print(f"P(Head or Tail) = {p_head_or_tail}")
```

## Non-Mutually Exclusive Events

For events that can overlap:

P(A ∪ B) = P(A) + P(B) - P(A ∩ B)

Example:

- A = Drawing a King
- B = Drawing a Heart

The King of Hearts belongs to both events.

```python
# King or Heart from a standard deck

total_cards = 52

kings = 4
hearts = 13
king_of_hearts = 1

p_king = kings / total_cards
p_heart = hearts / total_cards
p_king_and_heart = king_of_hearts / total_cards

p_king_or_heart = (
    p_king
    + p_heart
    - p_king_and_heart
)

print(f"P(King) = {p_king:.4f}")
print(f"P(Heart) = {p_heart:.4f}")
print(f"P(King and Heart) = {p_king_and_heart:.4f}")
print(f"P(King or Heart) = {p_king_or_heart:.4f}")
print(f"Percentage = {p_king_or_heart * 100:.2f}%")
```

### Why subtract the intersection?

If we calculate:

P(King) + P(Heart)

the King of Hearts is counted twice.

Therefore, we subtract:

P(King ∩ Heart)

once.

This gives the correct probability of drawing either a King or a Heart.

```python
def mutually_exclusive_probability(p_a, p_b):
    """Addition rule for mutually exclusive events."""
    return p_a + p_b


def non_mutually_exclusive_probability(p_a, p_b, p_a_and_b):
    """Addition rule for overlapping events."""
    return p_a + p_b - p_a_and_b


print(
    "Mutually Exclusive:",
    mutually_exclusive_probability(0.5, 0.5)
)

print(
    "Non-Mutually Exclusive:",
    non_mutually_exclusive_probability(
        4 / 52,
        13 / 52,
        1 / 52
    )
)
```

## Summary

- Probability measures the likelihood of an event.
- Probability lies between 0 and 1.
- Mutually exclusive events have no overlap.
- Non-mutually exclusive events have an intersection.
- For mutually exclusive events, add probabilities.
- For overlapping events, subtract the intersection once.
