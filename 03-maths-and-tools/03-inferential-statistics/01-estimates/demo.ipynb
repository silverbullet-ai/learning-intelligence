# Estimates — Inferential Statistics

This notebook demonstrates the basic concepts of:

- Population
- Sample
- Sample Mean
- Point Estimate
- Interval Estimate
- Margin of Error

The goal is to understand how sample data can be used to estimate an unknown population parameter.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
```

## 1. Create a Population

Let's create a hypothetical population representing the salaries of employees.

The population mean is known here for demonstration purposes, but in real-world inferential statistics, the population parameter is usually unknown.

```python
population = np.random.normal(
    loc=60000,
    scale=8000,
    size=10000
)

population_mean = population.mean()

print(f"Population size: {len(population)}")
print(f"Population mean: ₹{population_mean:,.2f}")
```

## 2. Take a Sample

Instead of using the entire population, we select a sample.

The sample mean will be used as a point estimate of the population mean.

```python
sample_size = 100

sample = np.random.choice(
    population,
    size=sample_size,
    replace=False
)

sample_mean = sample.mean()

print(f"Sample size: {sample_size}")
print(f"Sample mean: ₹{sample_mean:,.2f}")
```

## 3. Point Estimate

The sample mean is our point estimate of the population mean.

x̄ → Point Estimate of μ

```python
print(f"Population mean: ₹{population_mean:,.2f}")
print(f"Point estimate:  ₹{sample_mean:,.2f}")
print(f"Difference:      ₹{abs(population_mean - sample_mean):,.2f}")
```

## 4. Visualizing the Population and Sample

The sample is only a small portion of the population.

The sample mean provides a single-value estimate of the population mean.

```python
plt.figure(figsize=(10, 5))

plt.hist(population, bins=40, alpha=0.6, label="Population")
plt.axvline(
    population_mean,
    linestyle="--",
    linewidth=2,
    label="Population Mean"
)
plt.axvline(
    sample_mean,
    linestyle=":",
    linewidth=2,
    label="Sample Mean / Point Estimate"
)

plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.title("Population vs Point Estimate")
plt.legend()
plt.show()
```

## 5. Interval Estimate

A point estimate gives only one value.

An interval estimate provides a range.

For demonstration, we will construct an approximate interval using:

Point Estimate ± Margin of Error

```python
margin_of_error = 1500

lower_limit = sample_mean - margin_of_error
upper_limit = sample_mean + margin_of_error

print(f"Point Estimate:    ₹{sample_mean:,.2f}")
print(f"Margin of Error:   ₹{margin_of_error:,.2f}")
print(f"Lower Limit:       ₹{lower_limit:,.2f}")
print(f"Upper Limit:       ₹{upper_limit:,.2f}")
```

## 6. Visualizing the Interval

The interval can be represented as:

Point Estimate ± Margin of Error

```python
plt.figure(figsize=(10, 2))

plt.errorbar(
    sample_mean,
    0,
    xerr=margin_of_error,
    fmt="o",
    capsize=8,
    markersize=8
)

plt.axvline(
    population_mean,
    linestyle="--",
    linewidth=2,
    label="Population Mean"
)

plt.yticks([])
plt.xlabel("Salary")
plt.title("Interval Estimate")
plt.legend()

plt.show()
```

## 7. Repeated Sampling

A single sample can produce a sample mean that differs from the population mean.

Let's take many samples and calculate the mean of each sample.

This demonstrates why sampling matters in inferential statistics.

```python
sample_size = 100
number_of_samples = 1000

sample_means = []

for _ in range(number_of_samples):
    sample = np.random.choice(
        population,
        size=sample_size,
        replace=False
    )
    
    sample_means.append(sample.mean())

sample_means = np.array(sample_means)

print(f"Number of samples: {number_of_samples}")
print(f"Mean of sample means: ₹{sample_means.mean():,.2f}")
print(f"Population mean:      ₹{population_mean:,.2f}")
```

```python
plt.figure(figsize=(10, 5))

plt.hist(sample_means, bins=30, alpha=0.7)

plt.axvline(
    population_mean,
    linestyle="--",
    linewidth=2,
    label="Population Mean"
)

plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.title("Distribution of Sample Means")
plt.legend()

plt.show()
```

## 8. Key Observations

- The population parameter is the value we ultimately want to estimate.
- A sample provides a sample statistic.
- The sample mean is a point estimate of the population mean.
- A point estimate gives one numerical value.
- An interval estimate provides a range.
- Margin of error determines how far the interval extends around the point estimate.
- Different samples can produce different point estimates.

### Remember

Sample
→ Sample Statistic
