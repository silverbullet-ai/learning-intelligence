# AI vs ML vs DL vs Data Science

A short conceptual recap. Refer to `README.md` for detailed notes.

## 1. The hierarchy

```text
Artificial Intelligence (AI)
└── Machine Learning (ML)
    └── Deep Learning (DL)
```

Data Science overlaps with these fields; it is not another nested level.

## 2. Quick self-check

1. Which field uses multilayer neural networks?
2. Is every AI system necessarily based on ML?
3. Does Data Science only involve building ML models?
4. Which field learns patterns from data?

### Answers

1. Deep Learning.
2. No. AI includes approaches beyond ML.
3. No. Data Science also includes analysis, statistics, visualization, and other data-focused work.
4. Machine Learning.
