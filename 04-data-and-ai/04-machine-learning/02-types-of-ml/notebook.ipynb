# Types of Machine Learning

This notebook reviews the three core types of Machine Learning: Supervised, Unsupervised, and Reinforcement Learning. Detailed notes are in `README.md`.

## 1. Overview

```text
Machine Learning
├── Supervised Learning
│   ├── Regression
│   └── Classification
├── Unsupervised Learning
│   └── Clustering (covered in this lesson)
└── Reinforcement Learning
    └── Learning through actions and rewards
```

This is the introductory organization used in the lesson, not an exhaustive taxonomy of every ML method.

## 2. Supervised Learning

**Defining characteristic:** Input features are paired with known target/output values.

### Regression
- Predicts a continuous numerical value.
- Examples: house price, salary, temperature.
- Algorithms mentioned: Linear Regression, Ridge Regression, Lasso Regression, Elastic Net.

### Classification
- Predicts a category or class.
- **Binary classification:** Two classes, e.g. Pass/Fail.
- **Multiclass classification:** More than two classes.
- Algorithms mentioned: Logistic Regression, Decision Tree, Random Forest, AdaBoost, XGBoost, CatBoost.

Note: Decision Tree, Random Forest, AdaBoost, XGBoost, and CatBoost have regression and classification variants. Logistic Regression is commonly used for classification despite its name.

## 3. Unsupervised Learning

**Defining characteristic:** There is no known target/output feature supplied for the learning task.

### Clustering
- Discovers groups based on similarities in the data.
- Example: customer segmentation using salary and spending score.
- Algorithms mentioned: K-Means, Hierarchical Clustering, DBSCAN.

## 4. Reinforcement Learning

**Central idea:** An agent learns through actions and rewards or penalties received from an environment.

```text
Environment → Agent → Action → Environment
     ↑                              │
     └────── Reward / Penalty ──────┘
```

Core concepts:
- **Agent:** The learner or decision-maker.
- **Environment:** The world in which the agent operates.
- **Action:** A choice made by the agent.
- **Reward/Penalty:** Feedback received after an action.

The detailed Reinforcement Learning module is deferred to a later part of the course.

## 5. Compare the Three Types

| Aspect | Supervised | Unsupervised | Reinforcement |
|---|---|---|---|
| Learning signal | Labeled examples | Data without a supplied target | Rewards or penalties |
| Main idea | Learn to predict known outputs | Discover patterns or groups | Learn through actions and feedback |
| Typical task | Prediction | Clustering | Sequential decision-making |
| Example | House price prediction | Customer segmentation | Learning effective actions |
| Key concepts | Regression, Classification | Clustering | Agent, Environment, Action, Reward |

### Memory tricks
- **Supervised:** “I know the answer; learn to predict it.”
- **Unsupervised:** “I don't know the groups; find them.”
- **Reinforcement:** “Try actions; learn from rewards.”
