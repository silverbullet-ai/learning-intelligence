# Population and Sample

This notebook demonstrates:

- Population
- Sample
- Random Sampling
- Population vs Sample
- Sampling in Statistics

```python
import pandas as pd
import random
import matplotlib.pyplot as plt
import statistics
```

## Creating a Population

We'll create a population of 1000 students with randomly generated heights.

```python
population = [
    random.randint(150,190)
    for _ in range(1000)
]

len(population)
```

```python
population_df = pd.DataFrame({
    "Height": population
})

population_df.head()
```

## Visualizing the Population

```python
plt.figure(figsize=(8,4))

plt.hist(
    population,
    bins=15,
    edgecolor="black"
)

plt.title("Population Distribution")

plt.xlabel("Height (cm)")
plt.ylabel("Frequency")

plt.show()
```

## Taking a Random Sample

```python
sample = random.sample(
    population,
    100
)

len(sample)
```

```python
sample_df = pd.DataFrame({
    "Height": sample
})

sample_df.head()
```

## Sample Distribution

```python
plt.figure(figsize=(8,4))

plt.hist(
    sample,
    bins=10,
    edgecolor="black"
)

plt.title("Sample Distribution")

plt.xlabel("Height (cm)")
plt.ylabel("Frequency")

plt.show()
```

## Comparing Population and Sample Means

```python
print(
    "Population Mean:",
    round(statistics.mean(population),2)
)

print(
    "Sample Mean:",
    round(statistics.mean(sample),2)
)
```

## Population Size vs Sample Size

```python
labels = [
    "Population",
    "Sample"
]

sizes = [
    len(population),
    len(sample)
]

plt.figure(figsize=(5,5))

plt.pie(
    sizes,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Population vs Sample")

plt.show()
```

# Conclusion

Population

- Entire dataset
- Represented by N

Sample

- Smaller subset
- Represented by n

A representative sample allows us to estimate characteristics of the population while saving time, effort, and cost.
