# Types of Statistics

In this notebook we'll explore:

- Descriptive Statistics
- Inferential Statistics
- Population vs Sample
- Basic statistical summaries

```python
import pandas as pd
import matplotlib.pyplot as plt
import statistics
import random
```

```python
population = [
    180,170,162,150,160,
    172,168,175,178,165,
    169,171,182,177,166,
    174,173,181,167,176
]

population_df = pd.DataFrame({
    "Height": population
})

population_df
```

```python
population_df.describe()
```

```python
plt.figure(figsize=(7,4))

plt.hist(
    population,
    bins=6,
    edgecolor="black"
)

plt.title("Population Height Distribution")

plt.xlabel("Height (cm)")
plt.ylabel("Frequency")

plt.show()
```

```python
sample = random.sample(
    population,
    5
)

sample
```

```python
print("Mean :", statistics.mean(sample))
print("Median :", statistics.median(sample))
print("Variance :", statistics.variance(sample))
print("Standard Deviation :", statistics.stdev(sample))
```

```python
print(
    "Population Mean:",
    statistics.mean(population)
)

print(
    "Sample Mean:",
    statistics.mean(sample)
)
```

```python
labels = ["Population", "Sample"]

sizes = [
    len(population),
    len(sample)
]

plt.figure(figsize=(5,5))

plt.pie(
    sizes,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Population vs Sample Size")

plt.show()
```
